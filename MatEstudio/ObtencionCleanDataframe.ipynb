```python
import pandas as pd
import json
```

```python
df = pd.read_json("all_pokemon.json")

df.head(3)
```

```python
df.columns
```

```python
df_stats = pd.json_normalize(df["stats"])
df_types = pd.json_normalize(df["types"])

df_stats["id"] = df["id"]
df_types["id"] = df["id"]
```

```python
data = []

for indice in range(len(df["abilities"])):
    habilidades = df["abilities"][indice]
    id = df["id"][indice]
    if not habilidades:
        fila = {"hab_1": None, "id": id}
        data.append(fila)
        continue
    
    if len(habilidades) == 1:
        fila = {"hab_1": habilidades[0], "id": id}
        data.append(fila)

    if len(habilidades) == 2:
        fila = {"hab_1": habilidades[0],"hab_2": habilidades[1],"id": id}
        data.append(fila)

    if len(habilidades) == 3:
        fila = {"hab_1": habilidades[0],"hab_2": habilidades[1],"hab_3": habilidades[2],"id": id}
        data.append(fila)

df_habilidades = pd.DataFrame(data)

df_habilidades = df_habilidades[["hab_1","hab_2","hab_3","id"]]

df_habilidades
```

```python
df_stats
```

```python
df_types
```

```python
df_total = pd.merge(df_types,df_stats, on="id")
df_total = pd.merge(df_total,df_habilidades, on="id")

columnas = ["height","id","name","weight","ranking","popularity","base_experience"]

df = df[columnas]

df_total = pd.merge(df_total, df, on="id")

df_total
```

```python
df_total.duplicated().sum()

df_total = df_total.drop_duplicates()
```
