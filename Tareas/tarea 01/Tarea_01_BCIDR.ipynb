### 3.1 DataFrame unificado (1 punto)

Vamos a cargar en un DataFrame los datos de juegos guardados en el documento `bgg_GameItem.csv`. Al inspeccionar nuestros datos, podemos notar que los valores de ciertas comunas vienen "codificados" con ID. Para comenzar, vamos a juntar la información de los distintos archivos descargados.

**a) (0.8 pts)** Genere un DataFrame único con toda la información de cada juego, incluyendo: nombres de artistas y diseñadores involucrados, mecánicas, categorías, tipo, familia y editorial. Guarde este DataFrame en un nuevo archivo CSV.

````python
def crear_dataframe_unificado(df_item, df_type, df_category, df_family, df_mechanic, df_person, df_publisher):
    df_final = df_item.copy()

    # Función auxiliar para crear diccionarios con claves en str
    def dict_from(df, key_col, value_col):
        return dict(zip(df[key_col].astype(str), df[value_col]))
    

    dict_type      = dict_from(df_type, 'bgg_id', 'name')
    dict_person    = dict_from(df_person, 'bgg_id', 'name')
    dict_publisher = dict_from(df_publisher, 'bgg_id', 'name')
    dict_category  = dict_from(df_category, 'bgg_id', 'name')
    dict_mechanic  = dict_from(df_mechanic, 'bgg_id', 'name')
    dict_family    = dict_from(df_family, 'bgg_id', 'name')


    if 'game_type' in df_final.columns:
        df_final['game_type_name'] = df_final['game_type'].astype(str).map(dict_type)

    if 'designer' in df_final.columns:
        df_final['designer_name'] = df_final['designer'].astype(str).map(dict_person)

    def mapear_lista_ids(valor, diccionario):
        if pd.isna(valor) or str(valor).strip() == "":
            return None
        ids = str(valor).split(',')
        ids = [id_str.strip() for id_str in ids if id_str.strip() != ""]
        nombres = [diccionario[id_] if id_ in diccionario else f"Unknown (ID: {id_})" for id_ in ids]
        if len(nombres) == 0:
            return None
        return ", ".join(nombres)


    if 'artist' in df_final.columns:
        df_final['artist_names'] = df_final['artist'].apply(mapear_lista_ids, diccionario=dict_person)

    if 'publisher' in df_final.columns:
        df_final['publisher_names'] = df_final['publisher'].apply(mapear_lista_ids, diccionario=dict_publisher)

    if 'category_ids' in df_final.columns:
        df_final['category_names'] = df_final['category_ids'].apply(mapear_lista_ids, diccionario=dict_category)

    if 'mechanic_ids' in df_final.columns:
        df_final['mechanic_names'] = df_final['mechanic_ids'].apply(mapear_lista_ids, diccionario=dict_mechanic)

    # REORDENAR COLUMNAS
    columnas_orden = [
        'bgg_id', 'name', 'year', 'game_type_name', 'designer_name',
        'artist_names', 'publisher_names', 'min_players', 'max_players',
        'min_players_rec', 'complexity', 'language_dependency'
    ]
    if 'category_names' in df_final.columns:
        columnas_orden.append('category_names')
    if 'mechanic_names' in df_final.columns:
        columnas_orden.append('mechanic_names')

    otras_columnas = []
    for col in df_final.columns:
        if col not in columnas_orden:
            otras_columnas.append(col)
        columnas_finales = columnas_orden + otras_columnas

    df_final = df_final[columnas_finales]

    return df_final
````

````python

df_unificado = crear_dataframe_unificado(
    df_item=df_item,
    df_type=df_type,
    df_category=df_category,
    df_family=df_family,
    df_mechanic=df_mechanic,
    df_person=df_person,
    df_publisher=df_publisher
)
print("DATAFRAME UNIFICADO CREADO EXITOSAMENTE")
print("")
print(f"Dimensiones: {df_unificado.shape}")
print(f"Columnas principales: {df_unificado.columns.tolist()[:15]}...")  

print("\nPRIMERAS 5 FILAS:")
display(df_unificado.head())

# Guardamos el DF en .csv
df_unificado.to_csv("data/df_juegos_unificado.csv", index=False, encoding="utf-8") #UTF-8 por precaucion de carácteres especiales

````
