### 1.3 Inspección y limpieza

Inspeccione el DataFrame `penguins` y sus columnas. ¿Están correctamente asignados los tipos de cada columna? ¿Alguna podría ser categórica? ¿Existen valores nulos o inválidos? Aplique las transformaciones y limpiezas que considere necesarisas al DataFrame, explicando en los comentarios (o nuevas celdas) por qué hizo dichas transformaciones.

```python
import pandas as pd
```

```python
df1 = pd.read_csv('penguins/adelie.csv')
df2 = pd.read_csv('penguins/chinstrap.csv')
df3 = pd.read_csv('penguins/gentoo.csv')
```

```python
display(df1.head(), df2.head(), df3.head())
```

```python
#Agrego las especies en cada df por facilidad
df1['species'] = 'Adelie'
df2['species'] = 'Chinstrap'
df3['species'] = 'Gentoo'
```

```python
penguins = pd.concat([df1, df2, df3])
```

```python
display(penguins.head(), penguins.tail())
```

```python
#1.3
print("\nValores nulos por columna:")
penguins.isna().sum()
```

```python
print("Tipos de datos antes de la limpieza:")
print(penguins.dtypes)
```

```python
penguins['species'] = penguins['species'].astype('category')
penguins['sex'] = penguins['sex'].astype('category')
penguins['island'] = penguins['island'].astype('category')

#Elimino filas nulas
penguins_cleaned = penguins.dropna()

# Verificar los tipos de datos después de la limpieza
print("\nTipos de datos después de la limpieza:")
print(penguins_cleaned.dtypes)
```

```python
print("\nPrimeras filas del DataFrame limpio:")
print(penguins_cleaned.head())
```

### 2.2 Análisis agrupados

Responda las siguientes preguntas:

* a. En promedio, ¿cuál es la especie más pesada de pingüino?
* b. Para cada especie, ¿hay alguna diferencia significativa entre hembras y machos en relación al peso promedio?

```python
print('2.1A')
print(f'Número de entradas del dataframe limpio: {len(penguins_cleaned)}')
print('\n2.1B')
display(penguins_cleaned.describe().round(2))

print('\n2.1C')
nro_machos = len(penguins_cleaned[penguins_cleaned['sex'] == 'MALE'])
nro_hembras = len(penguins_cleaned[penguins_cleaned['sex'] == 'FEMALE'])
print(f"Machos: {nro_machos}")
print(f"Hembras: {nro_hembras}")
```

```python
print('2.2A')
promedios_peso = penguins_cleaned.groupby('species')['body_mass_g'].mean()


especie_mas_pesada = promedios_peso.idxmax()
mayor_peso = promedios_peso.max()

print("Peso promedio por especie:")
print(promedios_peso)
print(f"\nLa especie más pesada en promedio es {especie_mas_pesada} con un peso promedio de {mayor_peso} g.")
```

```python
prom_especie_sexo = penguins_cleaned.groupby(['species', 'sex'])['body_mass_g'].mean()

print("Peso promedio por especie y sexo:")
print(prom_especie_sexo)

print("\n"+"="*50)
#Comparo diferencias entre especies y sexo
for species in penguins_cleaned['species'].unique():
    peso_macho = prom_especie_sexo[species]['MALE']
    peso_hembra = prom_especie_sexo[species]['FEMALE']
    diff = abs(peso_macho - peso_hembra)
    print(f"\nEn la especie {species}:")
    print(f"- Machos: {peso_macho:.2f} g")
    print(f"- Hembras: {peso_hembra} g")
    print(f"- Diferencia: {diff.round(2)} g")
```

```python
print(f"Diferencia promedio: {round((674.66 + 411.76 + 805.09)/3, 2)}")
```

Si existe una diferencia significativa, los machos tienden a tener un peso mayor a las hembras con una diferencia promedio de 630.5 gramos 

```python
#Para la variacion, uso std (desviacion estandard)
std_ala_especie = penguins_cleaned.groupby('species')['flipper_length_mm'].std()

# Encontrar la especie con menor variación
especie_menor_variacion = std_ala_especie.idxmin()
least_variation = std_ala_especie.min()

print("\nDesviación estándar del tamaño de aleta por especie:")
print(std_ala_especie)
print(f"\nLa especie con menor variación en el tamaño de aleta es {especie_menor_variacion} con una desviación estándar de {round(least_variation, 2)} mm.")
```
