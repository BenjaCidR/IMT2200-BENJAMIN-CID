### (5 pts) 1.2 Limpieza básica de `gdpData`

Sobre `gdpData`:

1. Revise cuántos valores faltantes (`NaN`) hay por columna.  
2. Elimine filas donde **todas** las columnas de años (2020–2025) sean `NaN`.  
3. Para los valores faltantes que queden:
   - Use interpolación **por fila (por país)** a lo largo de los años (por ejemplo, usando `interpolate(axis=1)`),  
     y si aún quedan `NaN` en los extremos, reemplácelos por el valor más cercano disponible en el tiempo.
4. Transforme el DataFrame a formato **largo** (tidy) usando `melt`, de manera que quede con columnas:
   - `Country`, `Year` (int), `GDP` (float).
5. Asegúrese de que `Year` sea entero y `GDP` sea numérico.

Comente brevemente qué ventajas tiene el formato largo para el análisis.

---

```python
print(gdpData.isna().sum())
year_cols = ["2020", "2021", "2022", "2023", "2024", "2025"]

# Eliminar filas donde TODOS los años son NaN
gdpData = gdpData.dropna(subset=year_cols, how="all")
# Asegurar que las columnas de años sean numéricas (por si vienen como str)
gdpData[year_cols] = gdpData[year_cols].apply(pd.to_numeric, errors="coerce")

# Interpolar por FILA (axis=1), a lo largo de los años
gdpData[year_cols] = gdpData[year_cols].interpolate(axis=1)

# Si quedan NaN en los extremos, rellenar usando el valor más cercano en el tiempo
# Primero hacia la derecha (bfill), luego hacia la izquierda (ffill), todo por fila
gdpData[year_cols] = gdpData[year_cols].bfill(axis=1).ffill(axis=1)
gdp_long = gdpData.melt(
    id_vars="Country",          # columna que se mantiene fija
    value_vars=year_cols,       # columnas que “se derretirán”
    var_name="Year",            # nuevo nombre para la columna de años
    value_name="GDP"            # nuevo nombre para los valores
)

gdp_long["Year"] = gdp_long["Year"].astype(int)
gdp_long["GDP"] = pd.to_numeric(gdp_long["GDP"], errors="coerce")
gdp_long = gdp_long.dropna(subset=["GDP"])
display(gdp_long)
```
