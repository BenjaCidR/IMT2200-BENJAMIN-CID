## Pregunta 5 (2 pts)
Transforma los datos creando una nueva columna que clasifique a los Pokémon en "Livianos" y "Pesados" según si su peso está por debajo o por encima de la mediana.

```python
df_no_repe['weight'] = (df_no_repe['weight']).astype(float)
df_no_repe['weight'].sort_values(ascending=False)
df_no_repe[df_no_repe['weight'] == 9999]
```

```python
mean_weight = round(df_no_repe['weight'].mean(), 2)

#Utilizamos NumPy para hcaer la clasificación. Where de la forma (condition, [x, y], /), donde x es si se cumple la condición
df_no_repe['clasification_weight'] = np.where(
df_no_repe['weight'] < mean_weight, 'Liviano', 'Pesado'
)
df_no_repe.head()
```

```python
#Por curiosidad, se revisa la popularidad de pikachu
print(df_no_repe[df_no_repe['name'].str.lower() == 'pikachu']['popularity'])
display(df_no_repe['popularity'].sort_values(ascending=False))
mas_popular = df_no_repe[df_no_repe['popularity'] == "79.5%"]

#Se muestra el formato de esa columna.
display(mas_popular['name'])

#Como tiene un formato raro, se usa .iloc, indexador de pandas que te permite acceder a filas y columnas por su posición numérica, no por su etiqueta de índice.


print(f"Hay {(df_no_repe['popularity'] == "79.5%").sum()} pokemon(es) con el porcantaje de popularidad mayor.")
print(f"El pokemon más popula es {mas_popular['name'].iloc[0]}, con {mas_popular['popularity'].iloc[0]}.")

#tambien se podría hacer con .values()
print(f"El pokemon más popula es {mas_popular['name'].values()}, con {mas_popular['popularity'].values()}.")
```
