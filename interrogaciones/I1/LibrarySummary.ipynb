# PANDAS

La librería `pandas` es una de las herramientas más importantes en Python para la ciencia de datos y el análisis de datos. Su función principal es proporcionar estructuras de datos rápidas, flexibles y expresivas diseñadas para trabajar con datos "relacionales" o "etiquetados" de manera fácil e intuitiva.

Las dos estructuras de datos principales de pandas son:

1.  **`DataFrame`**: Es una tabla de datos bidimensional con columnas que pueden ser de diferentes tipos (números, texto, fechas, etc.), similar a una hoja de cálculo de Excel o una tabla de una base de datos SQL. Es la estructura más utilizada en pandas.

2.  **`Series`**: Es un arreglo unidimensional etiquetado, capaz de contener cualquier tipo de dato. Básicamente, es una sola columna de un `DataFrame`.

### ¿Qué puedes hacer con pandas?

*   **Leer y escribir datos**: Cargar datos desde y guardar datos hacia una gran variedad de formatos como CSV, Excel, JSON, SQL, HTML, y más.
    ````python
    # Leer un archivo CSV
    df = pd.read_csv('mi_archivo.csv')
    ````
*   **Limpieza y preparación de datos**:
    *   Manejar datos faltantes (eliminar o rellenar valores `NaN`).
    *   Filtrar filas y seleccionar columnas basadas en condiciones.
    *   Eliminar filas duplicadas.
    *   Cambiar los tipos de datos de las columnas.

*   **Análisis y exploración de datos**:
    *   Calcular estadísticas descriptivas (media, mediana, desviación estándar, etc.) con `.describe()`.
    *   Agrupar datos con `.groupby()` para realizar cálculos por categorías (por ejemplo, calcular las ventas promedio por ciudad).
    *   Contar valores únicos con `.value_counts()`.
    *   Ordenar los datos.

*   **Manipulación de datos**:
    *   Combinar, unir y concatenar múltiples `DataFrames`.
    *   Añadir o eliminar columnas.
    *   Transformar datos aplicando funciones a las columnas o filas.

*   **Trabajo con series de tiempo**: Tiene funcionalidades muy potentes para manipular y analizar datos indexados por fecha y hora.

En resumen, `pandas` es la herramienta fundamental para tomar datos en bruto (a menudo desordenados y de diferentes fuentes) y convertirlos en un formato limpio, ordenado y analizable, listo para el modelado o la visualización.

Aquí están las funciones y métodos más utilizados de `pandas`, agrupados por la tarea que realizan.

### 1. Carga y Guardado de Datos (I/O)

Estas funciones se usan para meter datos en un DataFrame y sacarlos.

*   `pd.read_csv()`: Lee datos desde un archivo CSV. Probablemente la función más usada de toda la librería.
*   `pd.read_excel()`: Lee datos desde un archivo de Excel.
*   `pd.read_json()`: Lee datos desde un archivo o cadena JSON.
*   `DataFrame.to_csv()`: Guarda el DataFrame en un archivo CSV.
*   `DataFrame.to_excel()`: Guarda el DataFrame en un archivo de Excel.

### 2. Inspección y Exploración Inicial

Una vez que tienes los datos, estas son las primeras funciones que llamas para entenderlos.

*   `DataFrame.head(n)`: Muestra las primeras `n` filas del DataFrame (por defecto 5).
*   `DataFrame.tail(n)`: Muestra las últimas `n` filas.
*   `DataFrame.info()`: Proporciona un resumen conciso del DataFrame, incluyendo el tipo de dato de cada columna, la cantidad de valores no nulos y el uso de memoria. Es fundamental para una primera revisión.
*   `DataFrame.describe()`: Genera estadísticas descriptivas para las columnas numéricas (media, desviación estándar, mínimo, máximo, cuartiles, etc.).
*   `DataFrame.shape`: Devuelve una tupla que representa las dimensiones (filas, columnas) del DataFrame.
*   `Series.value_counts()`: Cuenta la frecuencia de cada valor único en una columna (Series). Muy útil para columnas categóricas.
*   `DataFrame.columns`: Muestra los nombres de todas las columnas.

### 3. Selección y Filtrado de Datos

Para seleccionar subconjuntos de tus datos.

*   **Selección de columnas**: `df['nombre_columna']` (devuelve una Serie) o `df[['col1', 'col2']]` (devuelve un DataFrame).
*   `DataFrame.loc[]`: Accede a un grupo de filas y columnas por **etiquetas** (nombres de índice y columna). Es muy potente para el filtrado basado en condiciones.
    ````python
    # Selecciona filas donde la edad es mayor a 30 y solo las columnas 'nombre' y 'ciudad'
    df.loc[df['edad'] > 30, ['nombre', 'ciudad']]
    ````
*   `DataFrame.iloc[]`: Accede a los datos por su **posición** entera (índice numérico).
    ````python
    # Selecciona las primeras 5 filas y las primeras 2 columnas
    df.iloc[0:5, 0:2]
    ````

### 4. Limpieza de Datos

Para tratar con datos faltantes, duplicados o incorrectos.

*   `DataFrame.isnull()` o `isna()`: Detecta valores faltantes (devuelve un DataFrame booleano).
*   `DataFrame.dropna()`: Elimina filas o columnas con valores faltantes.
*   `DataFrame.fillna(valor)`: Rellena los valores faltantes con un `valor` específico.
*   `DataFrame.duplicated()`: Identifica filas duplicadas.
*   `DataFrame.drop_duplicates()`: Elimina filas duplicadas.
*   `DataFrame.astype(tipo)`: Cambia el tipo de dato de una o más columnas (por ejemplo, de `object` a `int` o `datetime`).

### 5. Agrupación y Agregación

Para resumir los datos.

*   `DataFrame.groupby('columna')`: Agrupa el DataFrame usando los valores de una o más columnas. Este método no hace nada por sí solo, se debe encadenar con una función de agregación.
*   Funciones de agregación usadas con `groupby()`:
    *   `.sum()`: Suma de los valores.
    *   `.mean()`: Media.
    *   `.count()`: Conteo de valores no nulos.
    *   `.size()`: Conteo de todos los elementos (incluyendo nulos).
    *   `.agg()`: Para aplicar múltiples funciones de agregación a la vez.
    ````python
    # Calcula las ventas promedio por ciudad
    df.groupby('ciudad')['ventas'].mean()
    ````

### 6. Manipulación y Transformación

Para modificar la estructura o los valores del DataFrame.

*   `DataFrame.drop()`: Elimina filas o columnas especificadas.
*   `DataFrame.apply(funcion)`: Aplica una función a lo largo de un eje (filas o columnas).
*   `pd.to_datetime()`: Convierte una columna a tipo de dato de fecha y hora, lo cual es muy potente para análisis de series de tiempo.
*   `DataFrame.sort_values()`: Ordena el DataFrame por los valores de una o más columnas.

Además de las funciones básicas, `pandas` tiene un conjunto muy potente de herramientas para tareas más avanzadas. Aquí hay otras funciones y métodos muy relevantes:

### 1. Combinación y Unión de DataFrames

Cuando tus datos están repartidos en varias tablas, estas funciones son esenciales.

*   `pd.merge()`: Combina dos DataFrames de forma similar a un `JOIN` de SQL. Puedes especificar cómo hacer la unión (inner, outer, left, right) y sobre qué columnas (claves) unir.
    ````python
    df_merged = pd.merge(df1, df2, on='id_columna', how='left')
    ````
*   `pd.concat()`: Concatena (apila) DataFrames uno encima del otro (eje 0) o uno al lado del otro (eje 1).
    ````python
    df_concatenado = pd.concat([df1, df2])
    ````
*   `DataFrame.join()`: Una forma más simple de combinar DataFrames, que funciona uniendo por el índice por defecto.

### 2. Remodelación de Datos (Pivoting)

Estas funciones te permiten cambiar la estructura de tu DataFrame de un formato "largo" a uno "ancho" y viceversa.

*   `DataFrame.pivot_table()`: Crea una tabla dinámica (como en Excel) a partir de tu DataFrame. Es extremadamente útil para resumir datos. Es una versión más potente de `groupby`.
    ````python
    # Crea una tabla con 'ciudad' en las filas, 'año' en las columnas y la media de 'ventas' como valores.
    pd.pivot_table(df, values='ventas', index='ciudad', columns='año', aggfunc='mean')
    ````
*   `DataFrame.melt()`: Es la operación inversa a `pivot`. Transforma un DataFrame de un formato ancho a uno largo. Esto es útil para hacer que los datos sean más "ordenados" (tidy data) para ciertos tipos de análisis o visualizaciones.

### 3. Manejo de Texto (String Methods)

Pandas ofrece un conjunto de métodos vectorizados para manipular texto en una Serie o columna a través del accesor `.str`.

*   `Series.str.contains('patrón')`: Comprueba si un patrón o subcadena está presente en cada elemento.
*   `Series.str.replace('antiguo', 'nuevo')`: Reemplaza ocurrencias de una cadena por otra.
*   `Series.str.split('delimitador')`: Divide las cadenas en un delimitador, a menudo creando nuevas columnas.
*   `Series.str.lower()` / `.str.upper()`: Convierte el texto a minúsculas o mayúsculas.
*   `Series.str.extract('regex')`: Extrae grupos de captura de una expresión regular como nuevas columnas.

### 4. Funciones de Series de Tiempo

Para datos que tienen un índice de fecha y hora, `pandas` es excepcionalmente potente.

*   **Accesor `.dt`**: Similar al accesor `.str`, el accesor `.dt` permite acceder a propiedades de fechas y horas de una Serie (que debe ser de tipo `datetime`).
    *   `Series.dt.year`, `Series.dt.month`, `Series.dt.day`, `Series.dt.weekday`: Extraen componentes individuales de la fecha.
*   `DataFrame.resample('frecuencia')`: Es una operación muy potente para cambiar la frecuencia de una serie de tiempo (por ejemplo, agrupar datos diarios en mensuales). Se usa de forma similar a `groupby`.
    ````python
    # Calcula las ventas medias mensuales a partir de datos diarios
    df_ventas_diarias.resample('M').mean()
    ````

### 5. Aplicación de Funciones

*   `DataFrame.transform(funcion)`: Es similar a `.apply()`, pero `transform` devuelve un objeto con la misma forma (mismos índices) que el original. Es muy útil dentro de un `groupby` para crear una nueva columna basada en una agregación del grupo.
    ````python
    # Calcula la media de ventas por ciudad y la asigna a cada fila correspondiente
    df['venta_media_ciudad'] = df.groupby('ciudad')['ventas'].transform('mean')
    ````
*   `DataFrame.pipe(funcion)`: Permite encadenar operaciones de una manera más limpia y legible, pasando el DataFrame como argumento a una serie de funciones.

Hay algunos aspectos más que son importantes y demuestran el poder de `pandas`:

### 1. El Tipo de Dato Categórico (`category`)

Ya lo mencioné brevemente, pero vale la pena profundizar. Cuando tienes una columna con un número limitado de valores repetidos (por ejemplo, una columna `país` o `género`), convertirla al tipo `category` es una práctica excelente.

````python
df['columna_repetida'] = df['columna_repetida'].astype('category')
````

**¿Por qué es tan útil?**

*   **Ahorro de Memoria:** Internamente, pandas no almacena la cadena de texto completa en cada fila. En su lugar, guarda una lista de las categorías únicas y luego usa números enteros (mucho más pequeños) para indicar a qué categoría pertenece cada fila. Con datasets grandes, el ahorro de memoria es masivo.
*   **Mejora del Rendimiento:** Las operaciones como `groupby` son significativamente más rápidas en columnas de tipo categórico que en columnas de tipo `object` (texto).

En tu notebook, la columna `eventType` es un candidato perfecto para esto, como se indica en la pregunta 1.3.6.

### 2. Visualización Integrada

Pandas se integra directamente con la librería de gráficos `matplotlib`, permitiéndote crear visualizaciones de forma muy rápida y sencilla directamente desde un DataFrame o una Serie usando el método `.plot()`.

*   Es ideal para una exploración rápida de los datos sin tener que importar y configurar otra librería.

````python
# Crear un histograma de una columna
df['edad'].plot(kind='hist', title='Distribución de Edades')

# Crear un gráfico de dispersión entre dos columnas
df.plot(kind='scatter', x='ingresos', y='gastos', title='Ingresos vs. Gastos')

# Graficar una serie de tiempo
df_ventas_temporada.plot(kind='line', figsize=(10, 5))
````

### 3. Funciones de Ventana Móvil (`rolling`)

Para datos ordenados (como series de tiempo), puedes realizar cálculos sobre una "ventana" deslizante de datos. Esto es fundamental para calcular medias móviles, desviaciones estándar móviles, etc.

*   `DataFrame.rolling(window=N)`: Crea un objeto de ventana móvil de tamaño `N`. Luego puedes encadenar una función de agregación (`.mean()`, `.sum()`, `.std()`).

````python
# Calcular la media móvil de 7 días para el precio de una acción
df['precio_media_7d'] = df['precio_cierre'].rolling(window=7).mean()
````

Estos tres puntos —el uso eficiente de tipos de datos, la visualización rápida y las capacidades analíticas avanzadas como las ventanas móviles— son los que realmente elevan a `pandas` de ser una simple herramienta de manipulación de tablas a un ecosistema completo para el análisis de datos en Python.

### `DataFrame.explode()`

El método `explode()` se utiliza para transformar un DataFrame que tiene columnas con valores de tipo lista (o tuplas, sets, etc.) en un nuevo DataFrame donde cada elemento de la lista tiene su propia fila. Los valores de las otras columnas se duplican para cada nueva fila creada.

Es la herramienta perfecta para "desanidar" o "desempaquetar" datos.

**¿Cuándo es útil?**

Imagina que tienes datos donde una sola fila representa una entidad, pero una de sus propiedades es una colección de elementos. Por ejemplo:
*   Un post de un blog con una lista de `tags`.
*   Una película con una lista de `géneros` o `actores`.
*   Un concierto con una lista de `obras` interpretadas.

Si quieres hacer un análisis sobre los elementos individuales (por ejemplo, contar cuántas veces se usa cada tag), necesitas que cada tag esté en su propia fila. Ahí es donde `explode()` es fundamental.

**Ejemplo práctico:**

Supongamos que tenemos este DataFrame:

````python
import pandas as pd

df = pd.DataFrame({
    'concierto_id': [101, 102],
    'director': ['Carrasco', 'Smith'],
    'obras': [['Sinfonía N°5', 'Obertura 1812'], ['Las cuatro estaciones']]
})

print("--- DataFrame Original ---")
print(df)
````
**Salida Original:**
```
--- DataFrame Original ---
   concierto_id  director                               obras
0           101  Carrasco  [Sinfonía N°5, Obertura 1812]
1           102     Smith           [Las cuatro estaciones]
```

Ahora, si queremos analizar cada obra por separado, usamos `explode()`:

````python
df_exploded = df.explode('obras')

print("\n--- DataFrame después de explode() ---")
print(df_exploded)
````
**Salida después de `explode()`:**
```
--- DataFrame después de explode() ---
   concierto_id  director                   obras
0           101  Carrasco          Sinfonía N°5
0           101  Carrasco         Obertura 1812
1           102     Smith  Las cuatro estaciones
```

Como puedes ver, la primera fila se duplicó para que cada obra de la lista tuviera su propia entrada. Ahora es trivial contar las obras, agrupar por director y obra, etc.

En el contexto de tu notebook, `explode()` sería una alternativa o un paso posterior a `json_normalize` si este último hubiera dejado alguna columna con listas que necesitaras desglosar. Es una herramienta clave para pasar de datos semi-estructurados (como JSON con listas anidadas) a un formato tabular plano y "ordenado" (tidy).

# NUMPY

`NumPy` (abreviatura de Numerical Python) es la librería fundamental para la computación científica y numérica en Python. Su principal objetivo es proporcionar un objeto de arreglo multidimensional de alto rendimiento y herramientas para trabajar con estos arreglos.

En esencia, `NumPy` introduce el objeto `ndarray` (N-dimensional array), que es una estructura de datos que permite almacenar y manipular grandes conjuntos de datos numéricos de manera muy eficiente.

### ¿Por qué es tan importante el `ndarray` de NumPy?

A diferencia de las listas de Python, que son flexibles pero lentas para operaciones matemáticas, los arreglos de NumPy son:

1.  **Más Rápidos:** Las operaciones matemáticas se realizan a través de código precompilado en C o Fortran, lo que las hace órdenes de magnitud más rápidas que usar bucles `for` en Python.
2.  **Más Eficientes en Memoria:** Almacenan los datos de manera más compacta que las listas de Python, ya que todos los elementos deben ser del mismo tipo.
3.  **Más Convenientes:** Permiten realizar operaciones matemáticas sobre arreglos completos sin necesidad de escribir bucles. Esto se conoce como **vectorización**.

**Ejemplo de Vectorización:**

Para multiplicar cada elemento de una lista por 2 en Python puro:
````python
lista_py = [1, 2, 3, 4, 5]
resultado_py = []
for item in lista_py:
    resultado_py.append(item * 2)
# resultado_py es [2, 4, 6, 8, 10]
````

Con NumPy:
````python
import numpy as np

array_np = np.array([1, 2, 3, 4, 5])
resultado_np = array_np * 2
# resultado_np es un array de NumPy: [ 2  4  6  8 10]
````

### ¿Qué puedes hacer con NumPy?

*   **Crear arreglos (`ndarray`)**: Crear arreglos de una o más dimensiones desde cero, a partir de listas, o usando funciones especializadas como `np.zeros`, `np.ones`, `np.arange`, `np.linspace`.
*   **Operaciones Matemáticas y Lógicas**: Realizar operaciones aritméticas elemento a elemento, así como aplicar funciones matemáticas complejas (`np.sin`, `np.sqrt`, `np.exp`) a arreglos completos.
*   **Manipulación de la Forma**: Cambiar las dimensiones de los arreglos (por ejemplo, de una lista larga a una matriz) con funciones como `reshape`.
*   **Indexación y Slicing Avanzado**: Seleccionar subconjuntos de datos de arreglos multidimensionales de formas muy potentes y flexibles.
*   **Álgebra Lineal**: Realizar operaciones de álgebra lineal como multiplicación de matrices, cálculo de determinantes, inversión de matrices, etc., a través del submódulo `np.linalg`.
*   **Estadística**: Calcular rápidamente estadísticas descriptivas como la media (`np.mean`), la desviación estándar (`np.std`), la suma (`np.sum`), el mínimo (`np.min`) y el máximo (`np.max`).
*   **Generación de Números Aleatorios**: Crear arreglos de números aleatorios con diferentes distribuciones, lo cual es fundamental para simulaciones y machine learning.

### El Pilar del Ecosistema Científico

`NumPy` no solo es potente por sí misma, sino que es la **base** sobre la que se construyen casi todas las demás librerías de ciencia de datos en Python, incluyendo:
*   **`pandas`**: Usa arreglos de NumPy internamente para almacenar los datos en sus `Series` y `DataFrames`.
*   **`matplotlib`**: Acepta arreglos de NumPy directamente para crear gráficos.
*   **`scikit-learn`**: La librería de machine learning más popular, que espera que los datos de entrada estén en formato de arreglos de NumPy.

En resumen, `NumPy` es la librería que dota a Python de la capacidad de realizar cálculos numéricos a alta velocidad, convirtiéndolo en un lenguaje de primera clase para el análisis de datos y la computación científica.

Aquí estan las funciones y características más utilizadas de `NumPy`, organizadas por su propósito.

### 1. Creación de Arreglos

Estas son las funciones para crear `ndarray` desde cero.

*   `np.array()`: La forma más básica. Convierte una lista o tupla de Python en un arreglo de NumPy.
    ````python
    import numpy as np
    mi_array = np.array([1, 2, 3])
    ````
*   `np.arange()`: Similar a la función `range()` de Python, pero devuelve un arreglo de NumPy en lugar de una lista.
    ````python
    array_secuencia = np.arange(0, 10, 2)  # [0 2 4 6 8]
    ````
*   `np.linspace()`: Crea un arreglo con un número específico de puntos, espaciados uniformemente entre un valor inicial y uno final. Muy útil para gráficos.
    ````python
    puntos_grafico = np.linspace(0, 10, 5)  # [ 0.   2.5  5.   7.5 10. ]
    ````
*   `np.zeros()` y `np.ones()`: Crean arreglos de una forma dada, llenos de ceros o unos.
    ````python
    matriz_zeros = np.zeros((2, 3))  # Matriz de 2x3 con ceros
    ````
*   `np.random.rand()`, `np.random.randn()`, `np.random.randint()`: Generan arreglos con números aleatorios (distribución uniforme, normal o enteros aleatorios, respectivamente).

### 2. Atributos para Inspeccionar Arreglos

No son funciones, sino atributos, pero se usan constantemente para entender la estructura de un arreglo.

*   `arr.shape`: Devuelve una tupla con las dimensiones del arreglo (ej: `(3, 4)` para una matriz de 3 filas y 4 columnas).
*   `arr.ndim`: El número de dimensiones del arreglo.
*   `arr.size`: El número total de elementos en el arreglo.
*   `arr.dtype`: El tipo de dato de los elementos del arreglo (ej: `float64`, `int32`).

### 3. Indexación y Selección (Slicing)

Esta es una de las características más potentes y utilizadas.

*   **Indexación básica**: `arr[5]`, `arr[1, 2]` (para acceder al elemento en la fila 1, columna 2).
*   **Slicing**: `arr[2:5]` (selecciona elementos del índice 2 al 4). Para arreglos multidimensionales: `arr[0:2, :]` (selecciona las primeras dos filas y todas las columnas).
*   **Indexación Booleana (Máscaras)**: Permite seleccionar elementos basados en una condición. Es extremadamente potente para filtrar.
    ````python
    arr = np.array([10, 20, 30, 40])
    mascara = arr > 25
    print(arr[mascara])  # Imprime [30 40]
    
    # O de forma más compacta:
    print(arr[arr > 25])
    ````

### 4. Operaciones Matemáticas (Vectorización)

El corazón de NumPy. Las operaciones se aplican a todo el arreglo a la vez, sin necesidad de bucles.

*   **Operadores aritméticos**: `+`, `-`, `*`, `/`, `**` funcionan elemento a elemento.
*   **Funciones Universales (ufuncs)**:
    *   `np.sqrt()`, `np.exp()`, `np.log()`
    *   Funciones trigonométricas: `np.sin()`, `np.cos()`

### 5. Funciones Estadísticas

Funciones de agregación que calculan una estadística sobre un arreglo.

*   `np.mean()`: Calcula la media.
*   `np.std()`: Calcula la desviación estándar.
*   `np.sum()`: Suma todos los elementos.
*   `np.min()` y `np.max()`: Encuentran el valor mínimo y máximo.
*   `np.argmin()` y `np.argmax()`: Encuentran el **índice** del valor mínimo y máximo.

Estas funciones también aceptan un parámetro `axis` para realizar el cálculo a lo largo de una fila (`axis=1`) o una columna (`axis=0`).

### 6. Manipulación de Arreglos

Funciones para cambiar la forma y estructura de los arreglos.

*   `arr.reshape()`: Cambia la forma del arreglo sin cambiar sus datos.
*   `arr.T` (o `arr.transpose()`): Transpone el arreglo (intercambia filas por columnas).
*   `np.concatenate()`: Une una secuencia de arreglos a lo largo de un eje existente.

### 7. Álgebra Lineal (`np.linalg`)

Un submódulo crucial para operaciones de álgebra lineal.

*   `np.dot()` o el operador `@`: Para la multiplicación de matrices.
*   `np.linalg.inv()`: Calcula la inversa de una matriz.
*   `np.linalg.eig()`: Calcula los autovalores y autovectores de una matriz.

# MATPLOTLIB

`matplotlib` es una de las librerías de visualización de datos más fundamentales y utilizadas en Python. Su propósito principal es crear una amplia gama de gráficos y diagramas estáticos, animados e interactivos de alta calidad.

En esencia, `matplotlib` te permite convertir tus datos (listas, arrays de NumPy, Series/DataFrames de pandas) en representaciones visuales.

### Características Principales:

1.  **Control Total:** Su mayor fortaleza es que te da un control granular sobre casi todos los aspectos de un gráfico: colores, grosores de línea, fuentes, etiquetas, leyendas, títulos, etc.

2.  **Variedad de Gráficos:** Puedes crear prácticamente cualquier tipo de gráfico 2D y también tiene soporte para gráficos 3D. Los más comunes incluyen:
    *   Gráficos de líneas (`plot`)
    *   Gráficos de dispersión (`scatter`)
    *   Gráficos de barras (`bar`, `barh`)
    *   Histogramas (`hist`)

# REQUESTS

La librería `requests` en Python se utiliza para enviar solicitudes HTTP de una manera sencilla y amigable. Permite que tu código interactúe con servicios web y APIs para descargar contenido, enviar datos, etc.

En resumen, te permite hacer cosas como:

*   **Descargar archivos** o el contenido de una página web (usando `requests.get()`).
*   **Enviar datos** a un servidor, por ejemplo, para rellenar un formulario (usando `requests.post()`).
*   Interactuar con APIs para obtener o modificar datos.

Es una de las librerías más populares de Python por su simplicidad para trabajar con la web.

Las funciones y métodos más utilizados de la librería `requests` se dividen en dos categorías: las que envían la solicitud y las que procesan la respuesta.

### 1. Para enviar solicitudes

Estas son las funciones que se usan para realizar los diferentes tipos de peticiones HTTP.

*   `requests.get(url, params=None, **kwargs)`: Para obtener datos de un servidor. Es la función más común.
    ````python
    respuesta = requests.get('https://api.github.com/events')
    ````
*   `requests.post(url, data=None, json=None, **kwargs)`: Para enviar datos a un servidor para crear un nuevo recurso (por ejemplo, enviar un formulario).
    ````python
    respuesta = requests.post('https://httpbin.org/post', json={'clave': 'valor'})
    ````
*   `requests.put(url, data=None, **kwargs)`: Para actualizar un recurso existente en el servidor.
    ````python
    respuesta = requests.put('https://httpbin.org/put', data={'clave': 'nuevo_valor'})
    ````
*   `requests.delete(url, **kwargs)`: Para solicitar que se elimine un recurso en el servidor.
    ````python
    respuesta = requests.delete('https://httpbin.org/delete')
    ````

### 2. Para procesar la respuesta

Una vez que haces una solicitud, obtienes un objeto `Response`. Los siguientes atributos y métodos de este objeto son los más usados:

*   `respuesta.status_code`: Devuelve el código de estado HTTP (por ejemplo, `200` para éxito, `404` para no encontrado). Es fundamental para comprobar si la solicitud fue exitosa.
    ````python
    if respuesta.status_code == 200:
        print('Solicitud exitosa')
    ````
*   `respuesta.text`: El contenido de la respuesta como una cadena de texto (Unicode). Útil para HTML, texto plano, etc.
    ````python
    print(respuesta.text)
    ````
*   `respuesta.content`: El contenido de la respuesta en bytes. Se usa para datos que no son texto, como imágenes o archivos PDF.
    ````python
    with open('imagen.png', 'wb') as f:
        f.write(respuesta.content)
    ````
*   `respuesta.json()`: Decodifica una respuesta en formato JSON y la convierte en un diccionario o lista de Python. Es extremadamente útil para trabajar con APIs.
    ````python
    datos = respuesta.json()
    ````
*   `respuesta.headers`: Un diccionario con las cabeceras HTTP de la respuesta.
    ````python
    print(respuesta.headers['Content-Type'])
    ````

# JSON

La librería `json` en Python es una librería estándar (viene incluida con Python) que te permite trabajar con datos en formato **JSON (JavaScript Object Notation)**.

JSON es un formato de texto ligero para el intercambio de datos. Es muy fácil de leer para los humanos y fácil de interpretar para las máquinas, lo que lo ha convertido en el estándar de facto para la comunicación con APIs web.

La librería `json` realiza dos tareas principales:

1.  **Codificación (Serialización):** Convertir objetos de Python (como diccionarios y listas) en una cadena de texto con formato JSON.
2.  **Decodificación (Análisis o Parsing):** Convertir una cadena de texto con formato JSON (o un archivo que la contenga) en objetos de Python equivalentes.

### Funciones más utilizadas

La librería tiene cuatro funciones principales que se dividen según si trabajan con **cadenas de texto (strings)** o con **archivos (files)**. La "s" al final del nombre de la función significa "string".

#### Para trabajar con Cadenas de Texto (en memoria):

*   `json.dumps(obj)`: **(dump-s)** Toma un objeto de Python (por ejemplo, un diccionario) y lo convierte en una **cadena de texto** con formato JSON.
*   `json.loads(s)`: **(load-s)** Toma una **cadena de texto** con formato JSON y la convierte en un objeto de Python.

#### Para trabajar con Archivos (en disco):

*   `json.dump(obj, fp)`: Toma un objeto de Python y lo escribe directamente en un **objeto de archivo** (`fp`, de file pointer).
*   `json.load(fp)`: Lee desde un **objeto de archivo** que contiene datos JSON y lo convierte en un objeto de Python.

### Ejemplo Práctico

````python
import json

# 1. Objeto de Python (un diccionario)
datos_python = {
    "nombre": "Juan Pérez",
    "edad": 30,
    "es_estudiante": False,
    "cursos": ["Python", "Ciencia de Datos"]
}

# --- Codificación (Python -> JSON) ---

# Convertir el diccionario a una cadena de texto JSON
json_string = json.dumps(datos_python, indent=4) # indent=4 lo hace más legible

print("--- Cadena JSON (resultado de dumps) ---")
print(json_string)

# Guardar el diccionario directamente en un archivo
with open('datos.json', 'w') as archivo:
    json.dump(datos_python, archivo, indent=4)

print("\nArchivo 'datos.json' creado.")


# --- Decodificación (JSON -> Python) ---

# Leer desde el archivo que acabamos de crear
with open('datos.json', 'r') as archivo:
    datos_leidos = json.load(archivo)

print("\n--- Objeto Python (resultado de load) ---")
print(datos_leidos)
print(f"El nombre es: {datos_leidos['nombre']}")
````

En resumen, la librería `json` es el puente que permite a tus programas de Python comunicarse fluidamente con sistemas que usan JSON, como APIs web, archivos de configuración y muchas otras aplicaciones modernas.

Las funciones más utilizadas de la librería `json` son cuatro, y se pueden agrupar en dos pares según si trabajan con cadenas de texto (`string`) o con archivos (`file`).

Un truco para recordarlas es que las funciones que terminan en **`s`** trabajan con **`s`trings**.

### 1. Para trabajar con Cadenas de Texto (en memoria)

*   `json.loads(json_string)`: **(load-s)**
    *   **Qué hace:** Decodifica (lee) una **cadena de texto** que contiene datos en formato JSON y la convierte en un objeto de Python (generalmente un diccionario o una lista).
    *   **Uso común:** Procesar la respuesta de una API que viene como texto.
    ````python
    import json
    
    json_texto = '{"nombre": "Ana", "edad": 25}'
    datos_python = json.loads(json_texto)
    
    print(datos_python['nombre'])  # Salida: Ana
    ````

*   `json.dumps(python_object)`: **(dump-s)**
    *   **Qué hace:** Codifica (escribe) un objeto de Python (un diccionario o lista) y lo convierte en una **cadena de texto** con formato JSON.
    *   **Uso común:** Preparar datos para enviarlos a una API en el cuerpo de una solicitud.
    ````python
    import json
    
    datos_python = {'nombre': 'Ana', 'edad': 25}
    json_texto = json.dumps(datos_python, indent=4) # indent hace el string legible
    
    print(json_texto)
    ````

### 2. Para trabajar con Archivos (en disco)

*   `json.load(file_pointer)`: **(load)**
    *   **Qué hace:** Lee desde un **archivo** que contiene datos JSON y lo convierte en un objeto de Python.
    *   **Uso común:** Cargar archivos de configuración o datasets guardados en formato `.json`.
    ````python
    import json
    
    # Suponiendo que existe un archivo 'datos.json'
    with open('datos.json', 'r') as archivo:
        datos_leidos = json.load(archivo)
    
    print(datos_leidos)
    ````

*   `json.dump(python_object, file_pointer)`: **(dump)**
    *   **Qué hace:** Escribe un objeto de Python directamente en un **archivo** en formato JSON.
    *   **Uso común:** Guardar el estado de un programa, datos o configuraciones en un archivo para uso futuro.
    ````python
    import json
    
    datos_para_guardar = {'id_usuario': 123, 'preferencias': ['oscuro', 'notificaciones']}
    
    with open('config.json', 'w') as archivo:
        json.dump(datos_para_guardar, archivo, indent=4)
    ````

### Resumen

| Función | Operación | Entrada | Salida |
| :--- | :--- | :--- | :--- |
| `json.loads()` | **Leer** desde **String** | Cadena de texto JSON | Objeto Python |
| `json.dumps()` | **Escribir** a **String** | Objeto Python | Cadena de texto JSON |
| `json.load()` | **Leer** desde **Archivo** | Puntero de archivo | Objeto Python |
| `json.dump()` | **Escribir** a **Archivo** | Objeto Python y puntero de archivo | Nada (escribe en el archivo) |

# BEAUTIFULSOUP4

`BeautifulSoup4` (generalmente importada como `bs4`) es una librería de Python diseñada para hacer **web scraping**, es decir, para extraer información de archivos HTML y XML.

No descarga las páginas web por sí misma (para eso se usa la librería `requests`), sino que su trabajo comienza una vez que ya tienes el contenido HTML. Su función principal es tomar ese texto HTML, que a menudo es desordenado y mal formado (conocido como "tag soup" o "sopa de etiquetas"), y convertirlo en un árbol de objetos de Python que es fácil de navegar y del cual se puede extraer datos.

### ¿Qué hace exactamente?

1.  **Parsea el HTML/XML:** Lee el documento HTML y lo interpreta, corrigiendo errores comunes para construir una estructura de datos lógica y coherente en forma de árbol.

2.  **Permite la Navegación:** Una vez que el HTML está parseado, puedes navegar por el árbol de etiquetas de manera muy intuitiva:
    *   Ir a etiquetas hijas.
    *   Ir a la etiqueta padre.
    *   Moverte entre etiquetas hermanas (las que están al mismo nivel).

3.  **Facilita la Búsqueda:** Proporciona métodos muy potentes para buscar etiquetas específicas basándose en diferentes criterios:
    *   Por el nombre de la etiqueta (ej: todos los `<a>` para los enlaces, o todos los `<h1>` para los títulos).
    *   Por sus atributos, como `class`, `id`, `href`, etc. (ej: encontrar la etiqueta `<div>` con `id="contenido-principal"`).
    *   Por el texto que contienen.
    *   Usando selectores CSS.

4.  **Permite la Extracción de Datos:** Una vez que encuentras la etiqueta que te interesa, puedes extraer fácilmente su contenido:
    *   El texto dentro de la etiqueta.
    *   El valor de cualquiera de sus atributos (por ejemplo, la URL del atributo `href` de un enlace).

### Ejemplo Práctico

Imagina que usaste `requests` para obtener el siguiente HTML y lo guardaste en una variable `html_doc`:

````html
<html>
<head><title>Mi Página Web</title></head>
<body>
    <p class="titulo"><b>Un título en negrita</b></p>
    <p class="historia">Había una vez...
        <a href="http://ejemplo.com/enlace1" class="hermana" id="link1">Enlace 1</a>,
        <a href="http://ejemplo.com/enlace2" class="hermana" id="link2">Enlace 2</a> y
        <a href="http://ejemplo.com/enlace3" class="hermana" id="link3">Enlace 3</a>;
    y vivieron felices para siempre.
    </p>
</body>
</html>
````

Con `BeautifulSoup`, puedes hacer esto:

````python
from bs4 import BeautifulSoup

# Suponiendo que html_doc contiene el HTML de arriba
html_doc = """...""" 

# 1. Crear el objeto 'soup' para parsear el HTML
soup = BeautifulSoup(html_doc, 'html.parser')

# 2. Navegar y buscar
titulo_pagina = soup.title.text  # Extrae el texto de la etiqueta <title>
print(f"El título es: {titulo_pagina}")

# Encontrar la primera etiqueta <b>
titulo_negrita = soup.find('b').text
print(f"El título en negrita es: {titulo_negrita}")

# Encontrar todos los enlaces (etiquetas <a>)
todos_los_enlaces = soup.find_all('a')
print(f"Se encontraron {len(todos_los_enlaces)} enlaces.")

# 3. Extraer datos de un enlace específico
primer_enlace = todos_los_enlaces[0]
url_primer_enlace = primer_enlace['href']  # Acceder al atributo 'href'
texto_primer_enlace = primer_enlace.text

print(f"El primer enlace apunta a: {url_primer_enlace}")
print(f"Y su texto es: {texto_primer_enlace}")
````

En resumen, `BeautifulSoup` es la herramienta que convierte el caos del código HTML en una estructura ordenada, permitiéndote buscar y extraer la información que necesitas de forma programática y sencilla.

Las funciones y métodos más utilizados de `BeautifulSoup` se centran en dos tareas principales: **encontrar etiquetas** y luego **extraer información** de ellas.

Aquí están las más importantes, siguiendo el flujo de trabajo típico de un scraper.

### 1. Creación del Objeto `Soup`

Este es siempre el primer paso. No es una función que se llame repetidamente, pero es la puerta de entrada a todo lo demás.

*   `BeautifulSoup(markup, 'parser')`: El constructor de la clase.
    *   `markup`: La cadena de texto con el contenido HTML o XML.
    *   `parser`: El motor que se usará para interpretar el `markup`. El más común y que viene por defecto es `'html.parser'`.
    ````python
    from bs4 import BeautifulSoup
    import requests
    
    respuesta = requests.get("http://ejemplo.com")
    soup = BeautifulSoup(respuesta.text, 'html.parser')
    ````

### 2. Funciones para Encontrar Etiquetas

Estas son las funciones que usarás el 90% del tiempo para localizar el contenido que te interesa.

*   `find(name, attrs, ...)`: Devuelve la **primera etiqueta** que coincide con los criterios de búsqueda. Si no encuentra nada, devuelve `None`.
    *   **Por nombre de etiqueta:** `soup.find('h1')`
    *   **Por atributo `id`:** `soup.find(id='contenido-principal')`
    *   **Por atributo `class`:** `soup.find('p', class_='parrafo-destacado')` (nota el guion bajo en `class_`, ya que `class` es una palabra reservada en Python).

*   `find_all(name, attrs, ...)`: Devuelve una **lista** con **todas las etiquetas** que coinciden con los criterios. Si no encuentra nada, devuelve una lista vacía. Es la función más potente y utilizada.
    *   **Todas las etiquetas de un tipo:** `soup.find_all('a')` (para obtener todos los enlaces).
    *   **Todas las etiquetas con una clase CSS:** `soup.find_all(class_='item-lista')`

*   `select(selector_css)`: Una alternativa muy popular a `find_all`. Permite usar **selectores CSS** para encontrar etiquetas, lo que es muy intuitivo para quienes tienen experiencia en desarrollo web. Siempre devuelve una **lista**.
    *   **Selector simple:** `soup.select('p')` (encuentra todos los párrafos).
    *   **Selector de clase:** `soup.select('.mi-clase')`
    *   **Selector de ID:** `soup.select('#mi-id')`
    *   **Selector complejo:** `soup.select('div#contenido > ul.lista-items > li')` (encuentra `<li>` dentro de un `<ul>` con clase `lista-items`, que a su vez está dentro de un `<div>` con id `contenido`).

*   `select_one(selector_css)`: Es el equivalente de `find` pero usando selectores CSS. Devuelve el **primer elemento** que coincide o `None`.

### 3. Métodos para Extraer Datos de una Etiqueta

Una vez que has encontrado una etiqueta (un objeto `Tag`) con las funciones anteriores, usas estos atributos y métodos para sacar la información.

*   `.text` (atributo): Devuelve el **contenido de texto** de una etiqueta y de todas sus etiquetas hijas, concatenado en una sola cadena.
    ````python
    parrafo = soup.find('p')
    print(parrafo.text) # 'Había una vez... Enlace 1, Enlace 2 y Enlace 3; y vivieron...'
    ````
*   `.get_text(strip=True)` (método): Similar a `.text`, pero es un método que ofrece más control. Usar `strip=True` es muy común para eliminar el exceso de espacios en blanco y saltos de línea al principio y al final.

*   **Acceso a atributos como un diccionario**: La forma más común de obtener el valor de un atributo (como `href` de un enlace o `src` de una imagen).
    ````python
    enlace = soup.find('a')
    url = enlace['href'] # Obtiene 'http://ejemplo.com/enlace1'
    
    imagen = soup.find('img')
    fuente_imagen = imagen['src']
    ````
*   `.get(attribute)` (método): Una forma más segura de acceder a los atributos. Devuelve `None` si el atributo no existe, en lugar de lanzar un error (`KeyError`).
    ````python
    # Si la etiqueta 'a' no tuviera el atributo 'href', enlace['href'] daría error.
    # Esto, en cambio, devolvería None.
    url = enlace.get('href') 
    ````

En resumen, el flujo de trabajo es: `BeautifulSoup()` para empezar, `find_all()` o `select()` para obtener una lista de etiquetas, un bucle `for` para iterar sobre ellas, y dentro del bucle, `.text` y `['attribute']` para extraer los datos que necesitas.

# GEOPANDAS

`GeoPandas` es una librería de Python que extiende las capacidades de la popular librería `pandas` para permitir el trabajo con **datos geoespaciales**.

En pocas palabras: **es pandas, pero para mapas**.

Toma la estructura de datos familiar y potente de `pandas` (el `DataFrame`) y le añade una capa de funcionalidad geoespacial. Esto te permite realizar análisis espaciales complejos de una manera muy intuitiva y "pythónica".

### ¿Qué hace exactamente?

1.  **Introduce Estructuras de Datos Geoespaciales:**
    *   **`GeoDataFrame`**: Es la estructura principal. Es idéntico a un `DataFrame` de `pandas`, pero con una columna especial adicional (por defecto llamada `geometry`). Esta columna no contiene números o texto, sino objetos geométricos (puntos, líneas, polígonos) que representan la ubicación de cada fila en el mapa.
    *   **`GeoSeries`**: Es la columna `geometry` por sí misma. Es una `Serie` de `pandas` que contiene los objetos geométricos.

2.  **Permite Leer y Escribir Formatos de Datos Espaciales:**
    *   Facilita enormemente la carga de datos desde formatos geoespaciales comunes como **Shapefile** (`.shp`), **GeoJSON**, y **GeoPackage**, convirtiéndolos directamente en un `GeoDataFrame`.
    *   También permite guardar tu `GeoDataFrame` de vuelta a estos formatos.

3.  **Realiza Operaciones Espaciales (Análisis Geoespacial):**
    *   **Uniones Espaciales (Spatial Joins):** Permite combinar dos `GeoDataFrames` basándose en su relación espacial. Por ejemplo, puedes tomar un `GeoDataFrame` de puntos (como la ubicación de tiendas) y unirlo con un `GeoDataFrame` de polígonos (como los barrios de una ciudad) para determinar en qué barrio se encuentra cada tienda.
    *   **Operaciones Geométricas:** Puedes calcular el área de polígonos, la longitud de líneas, la distancia entre geometrías, crear buffers (zonas de influencia) alrededor de puntos, etc.
    *   **Operaciones de Superposición (Overlay):** Puedes encontrar la intersección, la unión o la diferencia entre conjuntos de polígonos.

4.  **Facilita la Visualización de Mapas:**
    *   Se integra con `matplotlib` para permitir la creación de mapas de forma muy sencilla usando el método `.plot()`. Puedes colorear los polígonos de un mapa basándote en los valores de otras columnas (creando mapas coropléticos).

### 1. Lectura y Escritura de Datos

Estas son las funciones para cargar y guardar tus datos geoespaciales.

*   `geopandas.read_file(filename)`: Es la función más importante y el punto de partida de casi cualquier análisis. Lee una gran variedad de formatos de archivo geoespacial (como Shapefile, GeoJSON, GeoPackage) y los carga directamente en un `GeoDataFrame`.
    ````python
    import geopandas as gpd
    
    # Cargar un Shapefile del mundo
    mundo_gdf = gpd.read_file(gpd.datasets.get_path('naturalearth_lowres'))
    ````
*   `GeoDataFrame.to_file(filename, driver='...')`: El método complementario para guardar un `GeoDataFrame` en un archivo. Debes especificar el `driver` (formato), como `'ESRI Shapefile'` o `'GeoJSON'`.

### 2. Creación de un GeoDataFrame desde Cero

A menudo, tienes coordenadas (latitud/longitud) en un archivo no espacial, como un CSV.

*   `geopandas.points_from_xy(x, y)`: Una función auxiliar muy útil que toma dos columnas (o Series) de coordenadas X (longitud) e Y (latitud) y las convierte en una `GeoSeries` de objetos `Point`.
*   `geopandas.GeoDataFrame(dataframe, geometry=...)`: El constructor para crear un `GeoDataFrame`. Se usa un DataFrame de pandas existente y se le especifica cuál columna será la geometría.

**Ejemplo combinado:**
````python
import pandas as pd
import geopandas as gpd

# 1. Crear un DataFrame de pandas normal
data = {'Ciudad': ['Paris', 'Bogotá'], 'Latitud': [48.85, 4.71], 'Longitud': [2.35, -74.07]}
df = pd.DataFrame(data)

# 2. Crear la GeoSeries de puntos
geometria = gpd.points_from_xy(df.Longitud, df.Latitud)

# 3. Crear el GeoDataFrame
ciudades_gdf = gpd.GeoDataFrame(df, geometry=geometria)
````

### 3. Manejo de Sistemas de Coordenadas (CRS)

El CRS es fundamental para que los datos de diferentes fuentes se alineen correctamente.

*   `gdf.crs` (atributo): Permite **verificar** el Sistema de Coordenadas de Referencia actual de un `GeoDataFrame`.
*   `gdf.set_crs('epsg:xxxx')` (método): **Asigna** un CRS a un `GeoDataFrame` que no tiene uno definido. **No cambia la proyección**, solo la etiqueta.
*   `gdf.to_crs('epsg:xxxx')` (método): **Transforma (reproyecta)** las geometrías de un CRS a otro. Esta es la función que se usa para alinear datos o para cambiar a una proyección que permita medir áreas/distancias en metros.
    ````python
    # Asignar el CRS WGS84 (lat/lon) a nuestro GeoDataFrame de ciudades
    ciudades_gdf = ciudades_gdf.set_crs('epsg:4326')
    
    # Transformar el GeoDataFrame del mundo a una proyección de Mercator
    mundo_mercator = mundo_gdf.to_crs('epsg:3395')
    ````

### 4. Análisis Geoespacial

Aquí es donde ocurre la magia.

*   `geopandas.sjoin(gdf1, gdf2, how='inner', predicate='intersects')`: Realiza una **unión espacial (Spatial Join)**. Combina dos `GeoDataFrames` basándose en la relación espacial de sus geometrías (por ejemplo, qué puntos están dentro de qué polígonos).
    ````python
    # Unir las ciudades con los países para saber en qué país está cada ciudad
    ciudades_con_pais = gpd.sjoin(ciudades_gdf, mundo_gdf, how="inner", predicate="within")
    ````
*   `geopandas.overlay(gdf1, gdf2, how='intersection')`: Realiza operaciones de **superposición** entre dos capas de polígonos. `how` puede ser `'intersection'`, `'union'`, `'difference'`, etc.
*   **Atributos y métodos de la `GeoSeries` (`gdf.geometry`)**:
    *   `.area`: Calcula el **área** de cada geometría (polígono). Requiere un CRS proyectado para dar resultados en unidades como metros cuadrados.
    *   `.length`: Calcula la **longitud** de cada geometría (línea).
    *   `.buffer(distancia)`: Crea un polígono de **zona de influencia** (buffer) alrededor de cada geometría.
    *   `.intersects(otra_geometria)` / `.contains(otra_geometria)`: Devuelven `True` o `False` si una geometría se intersecta o contiene a otra.

### 5. Visualización

*   `GeoDataFrame.plot()`: El método más directo para crear un mapa. Es una extensión del `.plot()` de pandas y se basa en `matplotlib`.
    ````python
    # Crear un mapa simple del mundo
    mundo_gdf.plot()
    
    # Crear un mapa coroplético basado en la población
    mundo_gdf.plot(column='pop_est', legend=True)
    ````
